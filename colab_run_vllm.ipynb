## Étape 3 — Ollama (embeddings + contexte) et vLLM (génération)

Ollama reste utilisé uniquement pour `bge-m3` (embeddings, ingestion) et `gemma2:2b`
(contextualisation des chunks à l'ingestion) — pour garantir la compatibilité avec la
`vector_db/` versionnée dans le repo.

vLLM sert le modèle de génération finale : `gemma4:12b` côté Ollama devient
`google/gemma-4-12B-it-qat-w4a16-ct` côté HuggingFace — version quantifiée w4a16,
nécessaire pour tenir sur un GPU Colab (la version fp16 demande 40 Go+ de VRAM).

Le chargement de vLLM prend quelques minutes (téléchargement + compilation des kernels).

## Etape 4b - Interface Web Gradio (alternative interactive)

Lance cette cellule **a la place** de l'etape 4 batch pour poser des questions manuellement.
La reponse est affichee avec :
- Le texte de la reponse
- Le detail des chunks recuperes (source, page, score rerank)
- Pour chaque chunk : un **lien de telechargement** qui ouvre la page PDF avec le chunk **surligne en bleu**

Un lien public est genere automatiquement (valable 72h).

```python
import subprocess, time, os, sys, re, uuid
from pathlib import Path
from collections import defaultdict

!pip install -q gradio pymupdf

# Relance Ollama depuis /tmp (evite le bug llama-server specifique a Colab)
os.system("pkill -f ollama 2>/dev/null; pkill -f llama-server 2>/dev/null")
time.sleep(2)
subprocess.Popen(
    ["ollama", "serve"],
    cwd="/tmp",
    stdout=open("/tmp/ollama.log", "w"),
    stderr=subprocess.STDOUT,
)
time.sleep(5)

sys.path.insert(0, "/content/RAG_Aquila/src")
import ask
ask.VECTOR_DB_DIR = Path("/content/RAG_Aquila/vector_db")

# --- Bascule la génération sur vLLM (embeddings restent sur Ollama) ---
import requests
from langchain_openai import OpenAI as VLLMOpenAI

VLLM_PORT = 8000
for _ in range(60):
    try:
        if requests.get(f"http://localhost:{VLLM_PORT}/health", timeout=2).status_code == 200:
            break
    except Exception:
        pass
    time.sleep(5)
else:
    print("! vLLM ne répond pas sur le port 8000 — vérifier la cellule d'Étape 3 / /tmp/vllm.log")

ask.llm = VLLMOpenAI(
    model="gemma4-12b",                       # nom déclaré via --served-model-name à l'Étape 3
    base_url=f"http://localhost:{VLLM_PORT}/v1",
    api_key="vllm",                            # valeur fictive, vLLM ne vérifie pas la clé
    temperature=0,
    max_tokens=1024,
)

from agent import ask_question_agentic

import fitz          # PyMuPDF
import gradio as gr

DOCUMENTS_DIR = Path("/content/RAG_Aquila/documents")
# Dossier ou les PDF annotes sont ecrits sur disque pour etre servis par Gradio
# via /file=... (une vraie URL HTTP, contrairement aux Blob/data: URLs qui sont
# bloquees dans le contexte sandboxe de gr.HTML : URL.createObjectURL y est absent).
PDF_OUT_DIR = Path("/tmp/rag_aquila_highlighted")
PDF_OUT_DIR.mkdir(parents=True, exist_ok=True)


def _fragments(chunk_text: str) -> list[str]:
    """Extrait des fragments de texte propres pour la recherche dans le PDF."""
    text = re.sub(r"^\[.*?\]\n+", "", chunk_text.strip(), flags=re.DOTALL)
    text = re.sub(r"\*{1,2}([^*\n]+)\*{1,2}", r"\1", text)
    text = re.sub(r"^#{1,3}\s+", "", text, flags=re.MULTILINE)
    text = re.sub(r"\s+", " ", text).strip()
    words = text.split()
    frags = []
    for start in range(0, min(len(words), 40), 6):
        f = " ".join(words[start : start + 9])
        if len(f) > 20:
            frags.append(f)
    return frags


def _annotate_full_pdf_path(source: str, page_chunks: list) -> str:
    """
    Ouvre le PDF complet, surligne en bleu chaque chunk sur sa page, et ecrit le
    resultat dans PDF_OUT_DIR. Retourne le chemin absolu du fichier ecrit (ou ""
    si la source n'existe pas). Un seul appel par source = un seul PDF meme si
    plusieurs chunks viennent du meme doc. Nom de fichier unique (uuid) pour
    eviter qu'un navigateur garde une vieille version en cache entre deux questions.
    """
    pdf_path = DOCUMENTS_DIR / source
    if not pdf_path.exists():
        return ""
    doc = fitz.open(str(pdf_path))
    for page_num, chunk_text in page_chunks:
        page_idx = max(0, min(page_num - 1, len(doc) - 1))
        page = doc[page_idx]
        for frag in _fragments(chunk_text):
            areas = page.search_for(frag)
            if areas:
                for rect in areas:
                    annot = page.add_highlight_annot(rect)
                    annot.set_colors(stroke=(0.18, 0.51, 0.98))  # bleu
                    annot.update()
                break
    out_path = PDF_OUT_DIR / f"{uuid.uuid4().hex}_{source}"
    doc.save(str(out_path))
    doc.close()
    return str(out_path)


def repondre(question: str):
    reponse, chunks = ask_question_agentic(question, verbose=False)
    if not chunks:
        return reponse, "<p style='color:#888'>Aucun chunk recupere.</p>"

    # Regroupe les chunks par source : un seul PDF annote par document
    by_source = defaultdict(list)
    for chunk in chunks:
        src = chunk.metadata.get("source", "?")
        pg  = chunk.metadata.get("page", 1)
        by_source[src].append((pg, chunk.page_content))

    source_paths = {}
    for src, entries in by_source.items():
        out_path = _annotate_full_pdf_path(src, entries)
        if out_path:
            source_paths[src] = out_path

    # Chaque PDF annote est ecrit une fois par source et servi par Gradio via
    # /file=<chemin absolu> (voir la construction du lien plus bas).
    html = ["<div style='font-family:sans-serif;max-width:860px'>"]
    for i, chunk in enumerate(chunks, 1):
        source  = chunk.metadata.get("source", "?")
        page_num = chunk.metadata.get("page", 1)
        score   = chunk.metadata.get("rerank_score", None)
        preview = chunk.page_content[:380].replace("<", "&lt;").replace(">", "&gt;")
        if len(chunk.page_content) > 380:
            preview += "..."

        score_str = f" &mdash; score&nbsp;{score}" if score is not None else ""
        # Lien vers le fichier reel servi par Gradio (/file=...), pas de JS du
        # tout : ca fonctionne meme dans le contexte sandboxe de gr.HTML qui
        # bloque Blob/data: URLs (cause des echecs precedents).
        link = (
            f'<a href="/file={source_paths[source]}#page={page_num}"'
            f' target="_blank"'
            f' style="background:#2563eb;color:white;border:none;padding:5px 12px;'
            f'border-radius:5px;cursor:pointer;font-size:12px;white-space:nowrap;'
            f'text-decoration:none;display:inline-block">'
            f'&#128196; Ouvrir p.{page_num} surlignee</a>'
        ) if source in source_paths else ""

        html.append(
            f'<div style="border:1px solid #d1d5db;border-radius:8px;padding:12px 14px;'
            f'margin:8px 0;background:#f8fafc">'
            f'<div style="display:flex;justify-content:space-between;align-items:center;'
            f'margin-bottom:8px;gap:8px;flex-wrap:wrap">'
            f'<b style="color:#1e3a5f">Chunk {i} &mdash; {source}, page&nbsp;{page_num}{score_str}</b>'
            f'{link}</div>'
            f'<pre style="color:#374151;font-size:12px;margin:0;white-space:pre-wrap;'
            f'line-height:1.5">{preview}</pre>'
            f'</div>'
        )
    html.append("</div>")
    return reponse, "".join(html)


gr.Interface(
    fn=repondre,
    inputs=gr.Textbox(
        label="Ta question",
        lines=3,
        placeholder="Ex : Quelles sont les conditions pour obtenir le DENS ?",
    ),
    outputs=[
        gr.Textbox(label="Reponse", lines=10),
        gr.HTML(label="Chunks recuperes"),
    ],
    title="RAG Aquila - Mode interactif",
    description="Pose une question sur les formations ENS / Sorbonne Universite.",
    theme=gr.themes.Soft(),
).launch(share=True, allowed_paths=[str(PDF_OUT_DIR)])
```

## Étape 5 — Évaluation RAGAS

Lance cette étape **après** que `run_agentic_all.py` a terminé (étape 4).  
RAGAS évalue automatiquement les 5 métriques clés sur tous les résultats de `data/agentic_results.json`.

```python
# Installation de ragas + openai (requis par llm_factory pour pointer vers Ollama) + patch compatibilité langchain-community 0.4.x
!pip install -q ragas datasets openai

# langchain-community 0.4.x a supprimé ChatVertexAI — on patche l'import pour éviter le crash
import site, pathlib

# Cherche le fichier ragas/llms/base.py dans les packages installés
base_py = next(
    (p / "ragas" / "llms" / "base.py"
     for p in map(pathlib.Path, site.getsitepackages())
     if (p / "ragas" / "llms" / "base.py").exists()),
    None
)

if base_py:
    src = base_py.read_text(encoding="utf-8")
    # Applique le patch seulement si l'import direct est encore présent
    old = "from langchain_community.chat_models.vertexai import ChatVertexAI\nfrom langchain_community.llms import VertexAI"
    new = (
        "try:\n"
        "    from langchain_community.chat_models.vertexai import ChatVertexAI\n"
        "    from langchain_community.llms import VertexAI\n"
        "except ImportError:\n"
        "    ChatVertexAI = None\n"
        "    VertexAI = None"
    )
    if old in src:
        base_py.write_text(src.replace(old, new), encoding="utf-8")
        print("Patch appliqué sur ragas/llms/base.py")
    else:
        print("Patch déjà appliqué ou non nécessaire")
else:
    print("ragas/llms/base.py introuvable — vérifier l'installation")

print("Installation prête.")
```
